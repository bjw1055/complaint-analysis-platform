##Broadcasting
Broadcasting is a method for handling arrays of different shapes. It is fast because array operations are vectorized internally using C instead of Python. Even if the shapes of the two arrays differ , the array is processed if the operation conditions are met

```python
data = np.array([1.0,2.0])
data * 1.6
```



• np.mean() : A function that returns the arithmetic mean

• np.mean(s, axis = 0) : A function that returns the arithmetic mean along a column.

• np.mean(s, axis = 1) : A function that returns the arithmetic mean along the rows.



```python
s = [[10, 20, 30],
       [30, 40, 50],
       [50, 60, 70],
       [70, 80, 90]]

print("2차원 배열 :", s)
print("열에 따른 산술 평균 :", np.mean(s, axis=0))
print("행에 따른 산술 평균 :", np.mean(s, axis=1))
```
